```python
base = MQL.merge(DLS[['mql_id', 'won_date', 'seller_id']], on='mql_id', how='left')

base['has_won_date'] = base['won_date'].notna()
base['converted'] = base['has_won_date'].astype(int)

print('rows:     ', len(base))
print('converted:', base['converted'].sum())
```

```python
base['month'] = base['first_contact_date'].dt.strftime('%Y-%m')

leads_by_month = base.groupby('month').size()
conv_by_month  = base.groupby('month')['converted'].sum()
rate_by_month  = (100 * conv_by_month / leads_by_month).round(1)

monthly = pd.DataFrame({'Leads by Month': leads_by_month,
                        'Conv Month':  conv_by_month,
                        'Conv rate':  rate_by_month})
print(monthly)
```

```python
leads_won = base[base['converted'] == 1].copy()
leads_won['days_to_close'] = (leads_won['won_date'] - leads_won['first_contact_date']).dt.days

early = leads_won[leads_won['first_contact_date'] <  '2017-12-01']
later = leads_won[leads_won['first_contact_date'] >= '2017-12-01']

print('Early arrivals before Dec 2017')
print('count:', len(early))
print('fastest:', early['days_to_close'].min(),'days')
print('median:', early['days_to_close'].median(),'days')

print()
print('Later arrivals post Dec 2017')
print('count:', len(later))
print('fastest:', later['days_to_close'].min(),'days')
print('median:', later['days_to_close'].median(),'days')
```

```python
population = base[base['first_contact_date'] >= '2017-12-01'].copy()

print('leads:', len(population))
print('converted:', population['converted'].sum())
print('rate:', round(100 * population['converted'].mean(), 2), '%')
```

```python
population['origin_filled'] = population['origin'].fillna('Missing')

leads_o = population.groupby('origin_filled').size()
conv_o  = population.groupby('origin_filled')['converted'].sum()
rate_o  = (100 * conv_o / leads_o).round(1)

by_origin = pd.DataFrame({'leads': leads_o, 'conv': conv_o, 'rate': rate_o})
print(by_origin.sort_values('rate', ascending=False))

table = pd.crosstab(population['origin'], population['converted'])
result = stats.chi2_contingency(table)

print()
print('chi-square:', round(result[0], 1))
print('df:', result[2])
print('p-value:', result[1])
```

```python
page_counts = population['landing_page_id'].value_counts()

print('distinct pages:', len(page_counts))
one_lead = page_counts[page_counts == 1]
print('pages with 1 lead:', len(one_lead))
print('top 10 pages:', page_counts.head(10).sum(), 'leads')

leads_p = population.groupby('landing_page_id').size()
conv_p  = population.groupby('landing_page_id')['converted'].sum()

by_page = pd.DataFrame({'leads': leads_p, 'conv': conv_p})
by_page = by_page[by_page['leads'] >= 100]
by_page['rate'] = (100 * by_page['conv'] / by_page['leads']).round(1)

print()
print(by_page.sort_values('leads', ascending=False))
```
